```python
import pandas as pd
file_path = r"C:\Users\1\Downloads\Форма N-17 R29 всі відомства+приватні установи+МОЗ.xlsx"

xls = pd.ExcelFile(file_path)

xls.sheet_names
```

```python
df = pd.read_excel(
    file_path,
    sheet_name='F171000',
    header=None
)
#df.head(25)
```

```python
doctors = df.iloc[13:38, [1, 2]].copy()
```

```python
doctors.columns = ['region', 'doctors']
doctors = doctors.reset_index(drop=True)
doctors.index = doctors.index + 1
doctors.head()
```

```python
df1 = pd.read_excel(
    file_path,
    sheet_name='F171001',
    header=None
)
#df1.head(40)
```

```python
medical_staff=df1.iloc[11:36, [1, 2]].copy()
medical_staff.columns = ['region', 'medical_staff']
medical_staff =medical_staff.reset_index(drop=True)
medical_staff.index = medical_staff.index + 1
#medical_staff.head(25)
```

```python
df2 = pd.read_excel(
    file_path,
    sheet_name='F171003',
    header=None
)
#df2.head(10)
doc_pens=df2.iloc[10:35, [1, 2]].copy()
doc_pens.columns = ['region', 'doc_pensioner']
doc_pens =doc_pens.reset_index(drop=True)
doc_pens.index =doc_pens.index + 1
#doc_pens.head(25)
```

```python
df3 = pd.read_excel(
    file_path,
    sheet_name='F171004',
    header=None
)
#df3.head(10)
staff_pens=df3.iloc[10:35, [1, 2]].copy()
staff_pens.columns = ['region', 'staff_pensioner']
staff_penss =staff_pens.reset_index(drop=True)
staff_pens.index =staff_pens.index + 1
#staff_pens.head(25)
```

```python

doctors_rural=df2.iloc[39:64, [1, 2]].copy()
doctors_rural.columns = ['region', 'doc_rural']
doctors_rural =doctors_rural.reset_index(drop=True)
doctors_rural.index =doctors_rural.index + 1
#doctors_rural.head(25)
```

```python
medical_staff_rural=df3.iloc[39:64, [1, 2]].copy()
medical_staff_rural.columns = ['region', 'med_staff_rural']
medical_staff_rural =medical_staff_rural.reset_index(drop=True)
medical_staff_rural.index =medical_staff_rural.index + 1
#medical_staff_rural.head(25)
```

```python
doc_final = pd.merge(doctors, medical_staff, on='region', how='inner')

doc_final = pd.merge(doc_final, doc_pens, on='region', how='inner')

doc_final = pd.merge(doc_final, staff_pens, on='region', how='inner')
doc_final = pd.merge(doc_final, doctors_rural, on='region', how='inner')
doc_final = pd.merge(doc_final, medical_staff_rural, on='region', how='inner')
doc_final =doc_final.reset_index(drop=True)
doc_final.index =doc_final.index + 1
#print(doc_final)
doc_final
```
