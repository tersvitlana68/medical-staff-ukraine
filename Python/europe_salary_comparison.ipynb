# Порівняння оплати праці в країнах Європи

## Мета дослідження

Порівняти рівень оплати праці за основними видами економічної діяльності в окремих країнах Європи.

## Джерело даних

Eurostat. Mean annual earnings by sex, economic activity and collective pay agreement (2022).

```python
import pandas as pd

# Завантажуємо Excel-файл
data = pd.read_excel(
    'Mean annual earnings.xlsx',
    sheet_name='Data',
    header=None
)

# Збираємо дані з усіх блоків
rows = []

for start in range(0, len(data), 22):

    # Назва галузі
    sector = data.iloc[start + 8, 2]

    # Дані чотирьох країн
    for i in range(start + 14, start + 18):

        country = data.iloc[i, 0]
        salary = data.iloc[i, 1]

        if pd.notna(country) and pd.notna(salary):
            rows.append({
                'Країна': country,
                'Галузь': sector,
                'Рік': 2022,
                'Річна зарплата, EUR': salary
            })

# Створюємо чисту таблицю
eurostat = pd.DataFrame(rows)

display(eurostat)
```

```python
eurostat.to_csv(
    'eurostat_salary_2022.csv',
    index=False,
    encoding='utf-8-sig'
)
```
